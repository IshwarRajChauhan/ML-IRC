# EDA With Red Wine Data

## Data Set Information:

The two datasets are related to red and white variants of the Portuguese "Vinho Verde" wine. Due to privacy and logistic issues, only physicochemical (inputs) and sensory (the output) variables are available (e.g. there is no data about grape types, wine brand, wine selling price, etc.).

These datasets can be viewed as classification or regression tasks. The classes are ordered and not balanced (e.g. there are many more normal wines than excellent or poor ones). Outlier detection algorithms could be used to detect the few excellent or poor wines. Also, we are not sure if all input variables are relevant. So it could be interesting to test feature selection methods.

## Attribute Information:

### Input variables (based on physicochemical tests):

- 1 - fixed acidity
- 2 - volatile acidity
- 3 - citric acid
- 4 - residual sugar
- 5 - chlorides
- 6 - free sulfur dioxide
- 7 - total sulfur dioxide
- 8 - density
- 9 - pH
- 10 - sulphates
- 11 - alcohol

### Output variable (based on sensory data):

- 12 - quality (score between 0 and 10)

The easiest way to understand it:

> **EDA = Understand the data.**  
> **Feature Engineering = Improve/prepare the data for the ML model.**

## Difference

| EDA | Feature Engineering |
|---|---|
| Exploratory Data Analysis | Feature Engineering |
| Used to understand the dataset | Used to create/modify features |
| Finds patterns and problems | Makes features more useful for ML |
| Mostly analysis/visualization | Mostly transformation/creation |
| Doesn't necessarily change the dataset | Usually changes the dataset |
| Done before/during preprocessing | Done after understanding the data |

# Balanced vs Imbalanced Dataset

A **balanced dataset** is a dataset where the different classes have roughly similar numbers of samples.

An **imbalanced dataset** is a dataset where some classes have significantly more samples than others.

Balanced Dataset

For example:

```text
Quality 3 → 200
Quality 4 → 210
Quality 5 → 195
Quality 6 → 205
Quality 7 → 190
Quality 8 → 200

ImBalanced Dataset

For example:

Quality 3 → 10
Quality 4 → 53
Quality 5 → 577
Quality 6 → 535
Quality 7 → 167
Quality 8 → 17

````python
##Conclusion : It is an imbalanced datasets
df.quality.value_counts().plot(kind='bar')
plt.xlabel("Wine Quality")
plt.ylabel("Count")
plt.show()
````
